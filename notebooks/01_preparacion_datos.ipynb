# 1. Preparación de los datos

**Responsable:** Persona 1

Cargar el dataset `online_shoppers_intention.csv`, revisar sus columnas, separar la variable objetivo `Revenue`, convertir variables categóricas a numéricas y dividir en conjuntos de entrenamiento (80%) y prueba (20%).

**Fuente del dataset:** [sharmaroshan/Online-Shoppers-Purchasing-Intention](https://github.com/sharmaroshan/Online-Shoppers-Purchasing-Intention/blob/master/online_shoppers_intention.csv)

```python
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

DATA_PATH = Path("../data/online_shoppers_intention.csv")

df = pd.read_csv(DATA_PATH)
print(f"Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}")
df.head()
```

```python
print("Columnas del dataset:")
print(df.columns.tolist())

print("\nTipos de datos:")
print(df.dtypes)

print("\nValores nulos por columna:")
null_counts = df.isnull().sum()
print(null_counts)

if null_counts.sum() > 0:
    raise ValueError("El dataset contiene valores nulos. Revise la fuente antes de continuar.")
else:
    print("\nValidación OK: no hay valores nulos.")
```

```python
print("Resumen estadístico de columnas numéricas:")
df.describe()
```

```python
print("Distribución de la variable objetivo (Revenue):")
print(df["Revenue"].value_counts())

print("\nValores únicos en columnas categóricas:")
for col in ["Month", "VisitorType", "Weekend", "Revenue"]:
    print(f"  {col}: {df[col].unique()}")
```

```python
TARGET = "Revenue"

y = df[TARGET].map({"TRUE": 1, "FALSE": 0, True: 1, False: 0})
X_raw = df.drop(columns=[TARGET])

print(f"Variable objetivo (y): {y.name} — {y.value_counts().to_dict()}")
print(f"Variables predictoras (X): {X_raw.shape[1]} columnas")
```

```python
X = X_raw.copy()

X["Weekend"] = X["Weekend"].map({"TRUE": 1, "FALSE": 0, True: 1, False: 0})
X = pd.get_dummies(X, columns=["Month", "VisitorType"], drop_first=True)

print(f"Columnas después de codificación: {X.shape[1]}")
print(X.dtypes.value_counts())
X.head()
```

```python
RANDOM_STATE = 42
TEST_SIZE = 0.20

# ==============================================================================
# USO DE VARIABLES EN LAS SIGUIENTES FASES:
# ------------------------------------------------------------------------------
# X_train -> Se usa en fit() para que el modelo (Árbol o Random Forest) aprenda los patrones.
# y_train -> Se usa en fit() como las respuestas correctas para guiarlo durante el entrenamiento.
# X_test  -> Se usa en predict() para generar predicciones sin que el modelo las haya visto antes.
# y_test  -> Se compara con las predicciones para calcular Accuracy, Precision, Recall y la Matriz de Confusión.
# ==============================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y,
)

print(f"Entrenamiento: {X_train.shape[0]:,} muestras ({(1 - TEST_SIZE) * 100:.0f}%)")
print(f"Prueba:         {X_test.shape[0]:,} muestras ({TEST_SIZE * 100:.0f}%)")
print(f"\nProporción de compra (Revenue=1) en train: {y_train.mean():.3f}")
print(f"Proporción de compra (Revenue=1) en test:  {y_test.mean():.3f}")
```

## 📌 Guía para las siguientes fases

Los datos ya han sido limpiados, transformados y divididos. Para trabajar en los modelos de las siguientes fases, utilicen directamente las siguientes variables:

- **`X_train`**: Datos con las características del usuario para entrenar el modelo (80% del total, 26 columnas).
- **`y_train`**: Respuestas reales de compra usadas para entrenar el modelo (`1` = Compró, `0` = No compró).
- **`X_test`**: Datos reservados para hacer las predicciones de prueba (20% del total).
- **`y_test`**: Respuestas reales guardadas para evaluar qué tan acertadas fueron las predicciones.
